
function มีส่วนสำคัญทั้งหมด 4 ส่วน

1.   บอก python ว่าเราจะเขียนฟังก์ชั่น ชื่ออะไร  ```def function_name():``` (ขาดไม่ได้)
2.   กำหนดตัวแปรที่จะเป็น input  ```_Input_``` (ขาดได้)
3.   ส่วนประมวลผล ```do_something with _Input_ to get _Output_``` (ขาดไม่ได้)
4.   ส่วน output ```return _Output_```(ขาดได้)

# LOOP การวนซ้ำ

for เป็นคำที่ใช้บอก python ว่าเรากำลังเขียน loop
โดย for จะวนดึงสมาชิกจาก ```listA``` มาทำ process ```do_something```

```python
    for each_member in listA :

```

#### คำสั่ง for เป็นคำสั่งวนซ้ำที่ใช้ควบคุมการทำงานซ้ำๆ ในจำนวนรอบที่แน่นอน

#### เขียน function ตัดเกรด โดยที่ input เป็นคะแนน(0-100) , output เป็นเกรด (F-A)
#### วนลูปตัดเกรด input = [1,50,65,90,101,-5,49]

##### กำหนด
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ  0  แต่น้อยกว่า 50 จะได้เกรด F
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 50 แต่น้อยกว่า 55 จะได้เกรด D
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 55 แต่น้อยกว่า 60 จะได้เกรด D+
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 60 แต่น้อยกว่า 65 จะได้เกรด C
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 65 แต่น้อยกว่า 70 จะได้เกรด C+
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 70 แต่น้อยกว่า 75 จะได้เกรด B
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 75 แต่น้อยกว่า 80 จะได้เกรด B+
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 80  จะได้เกรด A
      แต่ว่าค่าคะแนนจะต้องไม่ต่ำกว่า 0 และมากกว่า 100 ไม่งั้นจะเกิด error

````python
# ฟังค์ชัน 1
def calculate_grade(score) :
  if score < 0 or score > 100 :
    print(f'{score} error')
  elif score >= 0 and score < 50 :
    print(f'คะแนนเท่ากับ {score} ได้เกรด F')
  elif score >= 50 and score < 55 :
    print(f'คะแนนเท่ากับ {score} ได้เกรด D')
  elif score >= 55 and score < 60 :
    print(f'คะแนนเท่ากับ {score} ได้เกรด D+')
  elif score >= 60 and score < 65 :
    print(f'คะแนนเท่ากับ {score} ได้เกรด C')
  elif score >= 65 and score < 70 :
    print(f'คะแนนเท่ากับ {score} ได้เกรด C+')
  elif score >= 70 and score < 75 :
    print(f'คะแนนเท่ากับ {score} ได้เกรด B')
  elif score >= 75 and score < 80 :
    print(f'คะแนนเท่ากับ {score} ได้เกรด B+')
  elif score >= 80 and score <= 100 :
    print(f'คะแนนเท่ากับ {score} ได้เกรด A')
````

````python
# ฟัง๕ืชัน 2
def calculate_grade2(score):
    if not 0 <= score <= 100:
        return f'{score} error'

    grades = ['F', 'D', 'D+', 'C', 'C+', 'B', 'B+', 'A']
    grade_index = max(0, min(7, (score - 45) // 5))
    grade = grades[grade_index]
    return f'คะแนนเท่ากับ {score} ได้เกรด {grade}'
````
